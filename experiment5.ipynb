```python
import math

# Evaluate board
def evaluate(board):
    for row in board:
        if row[0] == row[1] == row[2] != ' ':
            return 10 if row[0]=='O' else -10
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] != ' ':
            return 10 if board[0][col]=='O' else -10
    if board[0][0] == board[1][1] == board[2][2] != ' ':
        return 10 if board[0][0]=='O' else -10
    if board[0][2] == board[1][1] == board[2][0] != ' ':
        return 10 if board[0][2]=='O' else -10
    return 0

def is_moves_left(board):
    for row in board:
        if ' ' in row:
            return True
    return False

# -------- MINIMAX FUNCTION --------
def minimax(board, maximizingPlayer):
    score = evaluate(board)
    if score==10 or score==-10:
        return score
    if not is_moves_left(board):
        return 0
    
    if maximizingPlayer:
        best = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j]==' ':
                    board[i][j]='O'
                    best = max(best, minimax(board, False))
                    board[i][j]=' '
        return best
    else:
        best = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j]==' ':
                    board[i][j]='X'
                    best = min(best, minimax(board, True))
                    board[i][j]=' '
        return best

def find_best_move_minimax(board):
    bestVal = -math.inf
    bestMove = -1
    for pos in range(9):
        i,j = divmod(pos,3)
        if board[i][j]==' ':
            board[i][j]='O'
            moveVal = minimax(board, False)
            board[i][j]=' '
            if moveVal>bestVal:
                bestVal=moveVal
                bestMove=pos
    return bestMove

def print_board(board):
    for i in range(3):
        print(" | ".join(board[i]))
        if i<2:
            print("--+---+--")
    print()

# -------- DRIVER CODE --------
board = [' ']*9
board = [board[i*3:(i+1)*3] for i in range(3)]
print("Positions 0–8:")
print("0 | 1 | 2\n--+---+--\n3 | 4 | 5\n--+---+--\n6 | 7 | 8\n")
print_board(board)

while True:
    pos = int(input("Enter position (0-8) for X: "))
    i,j = divmod(pos,3)
    if board[i][j]!=' ':
        print("Cell occupied!")
        continue
    board[i][j]='X'
    print_board(board)
    
    if evaluate(board)==-10:
        print("X wins!")
        break
    if not is_moves_left(board):
        print("Tie!")
        break
    
    move=find_best_move_minimax(board)
    i,j=divmod(move,3)
    board[i][j]='O'
    print("AI (O) plays:")
    print_board(board)
    
    if evaluate(board)==10:
        print("O wins!")
        break
    if not is_moves_left(board):
        print("Tie!")
        break
```

```python
import math

def evaluate(board):
    for row in board:
        if row[0]==row[1]==row[2]!=' ':
            return 10 if row[0]=='O' else -10
    for col in range(3):
        if board[0][col]==board[1][col]==board[2][col]!=' ':
            return 10 if board[0][col]=='O' else -10
    if board[0][0]==board[1][1]==board[2][2]!=' ':
        return 10 if board[0][0]=='O' else -10
    if board[0][2]==board[1][1]==board[2][0]!=' ':
        return 10 if board[0][2]=='O' else -10
    return 0

def is_moves_left(board):
    for row in board:
        if ' ' in row:
            return True
    return False

# -------- ALPHA-BETA FUNCTION --------
def alphabeta(board, depth, alpha, beta, maximizingPlayer):
    score = evaluate(board)
    if score==10 or score==-10:
        return score
    if not is_moves_left(board):
        return 0
    if maximizingPlayer:
        best = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j]==' ':
                    board[i][j]='O'
                    best = max(best, alphabeta(board, depth+1, alpha, beta, False))
                    board[i][j]=' '
                    alpha=max(alpha,best)
                    if beta<=alpha:
                        break
        return best
    else:
        best = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j]==' ':
                    board[i][j]='X'
                    best=min(best,alphabeta(board,depth+1,alpha,beta,True))
                    board[i][j]=' '
                    beta=min(beta,best)
                    if beta<=alpha:
                        break
        return best

def find_best_move_alphabeta(board):
    bestVal = -math.inf
    bestMove = -1
    alpha=-math.inf
    beta=math.inf
    for pos in range(9):
        i,j=divmod(pos,3)
        if board[i][j]==' ':
            board[i][j]='O'
            moveVal=alphabeta(board,0,alpha,beta,False)
            board[i][j]=' '
            if moveVal>bestVal:
                bestVal=moveVal
                bestMove=pos
    return bestMove

def print_board(board):
    for i in range(3):
        print(" | ".join(board[i]))
        if i<2:
            print("--+---+--")
    print()

# -------- DRIVER CODE --------
board = [' ']*9
board = [board[i*3:(i+1)*3] for i in range(3)]
print("Positions 0–8:")
print("0 | 1 | 2\n--+---+--\n3 | 4 | 5\n--+---+--\n6 | 7 | 8\n")
print_board(board)

while True:
    pos = int(input("Enter position (0-8) for X: "))
    i,j=divmod(pos,3)
    if board[i][j]!=' ':
        print("Cell occupied!")
        continue
    board[i][j]='X'
    print_board(board)
    
    if evaluate(board)==-10:
        print("X wins!")
        break
    if not is_moves_left(board):
        print("Tie!")
        break
    
    move=find_best_move_alphabeta(board)
    i,j=divmod(move,3)
    board[i][j]='O'
    print("AI (O) plays:")
    print_board(board)
    
    if evaluate(board)==10:
        print("O wins!")
        break
    if not is_moves_left(board):
        print("Tie!")
        break
```
